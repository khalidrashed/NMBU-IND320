```python
# -----------------------------------------------------------------------------
# IND320 – Project Work, Part 1
# Jupyter Notebook: reading, cleaning, and plotting reservoirs.csv
# -----------------------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Nicer default plot style
plt.rcParams.update({
    "figure.figsize": (12, 5),
    "axes.grid": True,
    "grid.alpha": 0.3,
    "axes.titlesize": 12,
    "axes.labelsize": 10,
})

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)
```

```python
# -----------------------------------------------------------------------------
# 1) Read reservoirs.csv with Pandas
# Location: D2Dbook/data/reservoirs.csv (relative to notebook)
# -----------------------------------------------------------------------------

csv_path = "reservoirs.csv"
df_raw = pd.read_csv(csv_path)

print(f"Rows: {df_raw.shape[0]:,}  |  Columns: {df_raw.shape[1]}")
df_raw.head()
```

```python
# -----------------------------------------------------------------------------
# 2) Rename columns to English / understandable names
# -----------------------------------------------------------------------------
# Original (Norwegian)  → New (English)
#   dato_Id                     → date
#   omrType                     → area_type       ("EL", "NO", "VASS")
#   omrnr                       → area_nr         (1–5 for EL, 0 for aggregates)
#   iso_aar                     → iso_year
#   iso_uke                     → iso_week
#   fyllingsgrad                → fill_ratio      (0–1, fraction of capacity)
#   kapasitet_TWh               → capacity_TWh    (total reservoir capacity)
#   fylling_TWh                 → filled_TWh      (currently stored energy)
#   neste_Publiseringsdato      → next_publication
#   fyllingsgrad_forrige_uke    → fill_ratio_prev_week
#   endring_fyllingsgrad        → fill_ratio_change
# -----------------------------------------------------------------------------

rename_map = {
    "dato_Id": "date",
    "omrType": "area_type",
    "omrnr": "area_nr",
    "iso_aar": "iso_year",
    "iso_uke": "iso_week",
    "fyllingsgrad": "fill_ratio",
    "kapasitet_TWh": "capacity_TWh",
    "fylling_TWh": "filled_TWh",
    "neste_Publiseringsdato": "next_publication",
    "fyllingsgrad_forrige_uke": "fill_ratio_prev_week",
    "endring_fyllingsgrad": "fill_ratio_change",
}

df = df_raw.rename(columns=rename_map).copy()
df.columns.tolist()
```

```python
# Standard libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Make plots a bit larger and readable in the notebook
plt.rcParams["figure.figsize"] = (12, 4)
plt.rcParams["axes.grid"] = True
```

```python
# Load the reservoir data. The file is located in D2Dbook/data/reservoirs.csv
df = pd.read_csv("reservoirs.csv")

# Rename to English, understandable names
rename_map = {
    "dato_Id": "date",
    "omrType": "region_type",
    "omrnr": "region_number",
    "iso_aar": "iso_year",
    "iso_uke": "iso_week",
    "fyllingsgrad": "fill_fraction",
    "kapasitet_TWh": "capacity_TWh",
    "fylling_TWh": "fill_TWh",
    "neste_Publiseringsdato": "next_publication_date",
    "fyllingsgrad_forrige_uke": "fill_fraction_prev_week",
    "endring_fyllingsgrad": "fill_fraction_change",
}
df = df.rename(columns=rename_map)

# Parse dates (some are placeholder 0001-01-01)
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df["next_publication_date"] = pd.to_datetime(
    df["next_publication_date"], errors="coerce"
)

df.head()
```
